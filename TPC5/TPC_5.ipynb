### TPC5: Aplicação para gestão de alunos

Considere que o modelo do aluno e da turma têm a seguinte estrutura:

`aluno = (nome, id, [notaTPC, notaProj, notaTeste])`

`turma = [aluno]`

* Cria uma aplicação que coloca no monitor o seguinte menu de operações:
    - 1: Criar uma turma;
    - 2: Inserir um aluno na turma;
    - 3: Listar a turma;
    - 4: Consultar um aluno por id;
    - 8: Guardar a turma em ficheiro;
    - 9: Carregar uma turma dum ficheiro;
    - 0: Sair da aplicação
* No fim de executar a operação selecionada, a aplicação deverá colocar novamente o menu e pedir ao utilizador a opção para continuar;
* Utiliza a tua aplicação para criar uma turma com 5 alunos.

```python
def criarTurma():
    return []
 
 
def inserirAluno(listat):
    nome = input("Digite o nome do aluno: ")
    i = int(input("Digite o id do aluno: "))
    n1 = float(input("Digite a nota dos TPCs: "))
    n2 = float(input("Digite a nota do Projeto: "))
    n3 = float(input("Digite a nota do Teste: "))
    aluno = (nome, i, [n1, n2, n3])
    listat.append(aluno)
    return listat
 
 
def listarTurma(listat):
    print("Lista de alunos da turma:")
    for t in listat:
        print(f"O aluno {t[0]}, com o id {t[1]}, teve a nota {t[2][0]} no TPC, "
              f"{t[2][1]} no projeto e {t[2][2]} no teste.")
 
 
def idAluno(listat, i):
    for t in listat:
        if i == t[1]:
            print(f"O aluno cujo id é {i} é o/a {t[0]}.")
            return
    print("Aluno não encontrado.")
 
 
def guardarTurma(listat, fnome):
    with open(fnome, "w") as file:
        for nome, i, notas in listat:
            file.write(f"{nome};{i};{notas[0]}::{notas[1]}::{notas[2]}\n")
 
 
def recuperarTurma(fnome):
    listat = []
    with open(fnome, "r") as file:
        for linha in file:
            linha = linha.strip()
            if linha == "":
                continue
            nome, i, notas = linha.split(";")
            notas_lista = [float(n) for n in notas.split("::")]
            listat.append((nome, int(i), notas_lista))
    return listat
 
 
def menu():
    print("\n========= MENU =========")
    print("1: Criar uma turma")
    print("2: Inserir um aluno na turma")
    print("3: Listar a turma")
    print("4: Consultar um aluno por ID")
    print("8: Guardar a turma em ficheiro")
    print("9: Carregar uma turma dum ficheiro")
    print("0: Sair da aplicação")
    print("========================")
 
 
turma = []
modo = -1
while modo != 0:
    menu()
    modo = int(input("O que deseja efetuar? "))
    if modo == 0:
        print("Terminou")
    elif modo == 1:
        turma = criarTurma()
        print("Turma criada.")
    elif modo == 2:
        inserirAluno(turma)
    elif modo == 3:
        listarTurma(turma)
    elif modo == 4:
        ident = int(input("Qual id deseja procurar: "))
        idAluno(turma, ident)
    elif modo == 8:
        guardarTurma(turma, "turma.txt")
        print("A turma foi guardada no ficheiro turma.txt")
    elif modo == 9:
        turma = recuperarTurma("turma.txt")
        print("Turma carregada.")
    else:
        print("Tente outro número")
```
