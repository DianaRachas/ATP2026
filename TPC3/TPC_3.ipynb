### TPC3: Corrida para o 100

O total começa em 0. O jogador e o computador alternam somando um número de 1 a 10 ao total. Quem atingir exatamente o número 100 vence.
- Se o computador jogar primeiro deverá vencer;
- Se o computador jogar em segundo poderá vencer se o outro jogador fizer uma aposto fora da estratégia vencedora.

Implementa o jogo com 2 vertentes: o computador joga primeiro (deverá ganhar sempre), e o computador joga em segundo lugar (poderá ganhar ou não dependendo das jogadas do outro).

```python
def corrida100():
    escolha=input("Quem começa computador ou utilizador: ")
    print("Quem chegar primeiro aos 100 VENCE")
    if escolha=="computador":
        total=1
        while total < 100:

            if total < 12:
                print(f"O total é {total} tirei " + str(1))
                op2 = int(input("Queres retirar quanto? (1-10): "))
                print(f"O total é {total} tiraste {op2}")
                total+=op2
                op1= 12 - total
                total+=op1
                print(f"O total é {total} tirei {op1}")

            else:
                op2 = int(input("Queres retirar quanto? (1-10): "))
                total+=op2
                print(f"O total é {total} tiraste {op2}")
                op1= 11 - op2
                total+=op1
                print(f"O total é {total} tirei {op1}")
        print("Se fôr eu o primeiro a jogar ganho-te sempre:) É melhor começares tu hahahaha")

    if escolha=="utilizador":
            total=0
            op=0

            while total < 100:
                op1 = int(input("Queres retirar quanto? (1-10): "))
                total+=op1
                print(f"O total é {total} tiraste {op1}")

                if total < 100:

                    if op1 == 1:
                        op2= 11 - op1
                        total+=op2
                        print(f"O total é {total} tirei {op2}")

                    elif op == 0:
                        op = 12-op1
                        total+=op
                        print(f"O total é {total} tirei {op}")

                    else:
                        op2 = 11-op1
                        total+=op2
                        print(f"O total é {total} tirei {op2}")
            print("Se fôr o segundo a jogar só ganho se me deixares:)")            

corrida100()            
```
